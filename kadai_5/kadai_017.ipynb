```python
class Human:
    def __init__(self,name,age):
        self.name = ""
        self.name = name
        self.age = age
    def check_adult(self):
        if self.age >= 20 :
            print("adult")
        else:
            print("child")

user_list = [
    Human("ryo",33),
    Human("rina",30),
    Human("kohana",0)
]

for user in user_list:
    user.check_adult()
```
